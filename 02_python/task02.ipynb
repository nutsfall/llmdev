## 仕様

入力された税抜価格をもとに、消費税込の価格と送料を合計した「送料込価格」を計算し、結果を表示するソースコードを作成してください。

- 税込価格が `2000円以上` の場合は送料無料
- 税込価格が `2000円未満` の場合は送料 `350円`
- 消費税率は `10%`
- 税抜価格の入力には `input()` 関数を使用
- 結果の表示には `f文字列` を使用

## 実行結果の例

### 税抜価格が `2000円` の場合

```text
送料は無料です
送料込税込価格は2200円です
```

### 税抜価格が `1500円` の場合

```text
送料は350円です
送料込税込価格は2000円です
```

````python
# 税抜価格を入力
price_excl_tax = int(input("税抜価格を入力してください: "))

# 税込価格を計算（消費税10%）
price_incl_tax = int(price_excl_tax * (100+10)/100)

# 送料を判定
if price_incl_tax >= 2000:
    shipping_fee = 0
    print("送料は無料です")
else:
    shipping_fee = 350
    print(f"送料は{shipping_fee}円です")

# 送料込税込価格を計算して表示
total_price = price_incl_tax + shipping_fee
print(f"送料込税込価格は{total_price}円です")
````

## 仕様

ユーザーに2つの数値を入力させ、それらを割り算するプログラムを作成してください。  
以下の条件を満たしてください。

- 数値の代わりに文字を入力した場合は、`エラー: 数値を入力してください` と表示する  
- 割る数が `0` の場合は、`エラー: 0で割ることはできません` と表示する  
- 計算が正常に完了した場合は、結果を表示する

## 実行結果の例

1つ目の数を `100`、2つ目の数を `10` と入力した場合

```text
計算結果: 10.0
```

1つ目の数を `100`、2つ目の数を `0` と入力した場合

```text
エラー: 0で割ることはできません
```

1つ目の数を `a` と入力した場合

```text
エラー: 数値を入力してください
```

1つ目の数を `100`、2つ目の数を `b` と入力した場合

```text
エラー: 数値を入力してください
```

````python
try:
    number_one = int(input("数字1: "))
    number_two = int(input("数字2: "))
    result = number_one / number_two
except ZeroDivisionError:
    print("エラー: 0で割ることはできません")
except ValueError:
    print("エラー: 数値を入力してください")
else:
    print(f"計算結果: {result:.1f}")
````

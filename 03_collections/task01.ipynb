# 課題：成績リストの集計

## 仕様
以下の成績リストを使います。

```python
grades = [85, 90, 78, 92, 88]
```

このリストを使って、次を求めてください。

1. 平均点  
2. 最高点と最低点  
3. 平均以上の学生数  

最後に結果を出力してください。

## 実行結果の例

成績リストが `[85, 90, 78, 92, 88]` の場合

```text
平均点: 86.6
最高点: 92
最低点: 78
平均以上の学生数: 3
```

成績リストが `[70, 75, 80, 85, 90]` の場合

```text
平均点: 80.0
最高点: 90
最低点: 70
平均以上の学生数: 3
```

````python
grades = [85, 90, 78, 92, 88]
# grades = [70, 75, 80, 85, 90]

max_score = grades[0]
min_score = grades[0]

# avarage 
total = 0
count = 0
for g in grades:
    total += g
    count += 1
    if g > max_score:
        max_score = g
    if g < min_score:
        min_score = g

avarage = total / count

# avarage 以上の人数
students_over_avarage = 0
for g in grades:
    if g >= avarage:
        students_over_avarage  += 1

# 結果の表示
print(f"平均点: {avarage:.1f}")
print(f"最高点: {max_score:d}")
print(f"最低点: {min_score:d}")
print(f"平均点以上の人数: {students_over_avarage:d}")
````

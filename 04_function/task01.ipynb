## 仕様

### 1. 関数の仕様

文字列を受け取り、バリデーション結果を `bool` で返す関数を作成する。

#### バリデーションルール

1. ユーザー名は **5文字以上15文字以内**  
   - 条件を満たさない場合：  
     `エラー: ユーザー名は5文字以上15文字以内で入力してください` を表示し、`False` を返す

2. ユーザー名に **空白は使用不可**  
   - 空白が含まれる場合：  
     `エラー: ユーザー名に空白を含めることはできません` を表示し、`False` を返す

3. ユーザー名は **英数字のみ使用可能**（アルファベットと数字）  
   - それ以外の文字が含まれる場合：  
     `エラー: ユーザー名は英数字のみ使用できます` を表示し、`False` を返す

4. すべての条件を満たす場合  
   - `True` を返す（成功メッセージは表示しない）

### 2. 呼び出し側の仕様

1. ユーザーにユーザー名を入力させる  
2. 入力されたユーザー名を、上記のバリデーション関数で判定する  
3. 判定結果に応じて以下を表示する  
   - `True` の場合：`ユーザー名は有効です`  
   - `False` の場合：`不正なユーザー名です`

## 実行結果の例

ユーザー名を JohnDoe123 と入力した場合

```ユーザー名は有効です```

ユーザー名を JD と入力した場合

```
エラー: ユーザー名は5文字以上15文字以内で入力してください
不正なユーザー名です
```

ユーザー名を John@Doe と入力した場合

```
エラー: ユーザー名は英数字のみ使用できます
不正なユーザー名です
```

ユーザー名を John Doe と入力した場合

```
エラー: ユーザー名に空白を含めることはできません
不正なユーザー名です
```

````python
def validate_name(name):
    try:
        if " " in name:
            raise ValueError("ユーザー名に空白を含めることはできません")
        if not (5 <= len(name) <= 15):
            raise ValueError("ユーザー名は5文字以上15文字以内で入力してください")
        if not name.isalnum():
            raise ValueError("ユーザー名は英数字のみ使用できます")
    except ValueError as e:
        print(f"エラー: {e}")
        return False

    return True

input_name = input("ユーザー名を入力してください: ")

if validate_name(input_name):
    print("ユーザー名は有効です")
else:
    print("不正なユーザー名です")
````
